```python
import os
import sys
from qmiotools.integrations.qiskitqmio import FakeQmio,QmioBackend
from qiskit.visualization import plot_gate_map, plot_coupling_map, plot_error_map
from qmiotools.data import QUBIT_POSITIONS

%matplotlib inline
```

```python
#Parameter isle choose between 'full','isle1', 'isle2', 'isle3'

isle='isle2'

backend=QmioBackend(logging_level=3,qmio_partition=isle)
#fakebackend=FakeQmio(qmio_partition='isle1')

isle_qbit_positions=[]
for key in backend._isle_to_log_mapping:
    isle_qbit_positions.append(QUBIT_POSITIONS[int(key[2:-1])])

#print(isle_qbit_positions)
    
plot_error_map(backend, qubit_coordinates=isle_qbit_positions)
```

```python
from qiskit import QuantumCircuit, transpile
import matplotlib.pyplot as plt
from qiskit.converters import circuit_to_dag

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure(0,0)
qc.measure(1, 1)
qct = transpile(qc, backend)

qc.draw('mpl')


```

```python
compiled_circuit = transpile(qc, backend)
compiled_circuit.draw("mpl")
```

```python
# Execute the circuit


result = backend.run(compiled_circuit,output_qasm3=True).result()

# Get measurement counts
counts = result.get_counts(qc)
print(counts)
```
